# 01 — Data Audit

## Objective

This notebook audits the original healthcare datasets before cleaning or analysis.

The audit will:

- Verify the available raw data files.
- Inspect columns, dimensions, and data types.
- Identify missing values and duplicate rows.
- Review the years and measurement units.
- Preserve the original raw files without modification.

## 1. Setup and File Discovery

Import the required libraries and locate the raw data files.

```python
from pathlib import Path
import pandas as pd

project_root = Path.cwd()

if project_root.name == "notebooks":
    project_root = project_root.parent

raw_dir = project_root / "data" / "raw"

print("Project root:", project_root)
print("Raw data directory:", raw_dir)
print("Directory exists:", raw_dir.exists())
```

```python
csv_files = sorted(raw_dir.glob("*.csv"))

print("Number of CSV files:", len(csv_files))

for file in csv_files:
    print("-", file.name)
```

## 2. Load Raw Datasets

Load the original datasets into pandas DataFrames without modifying the source files.

```python
inpatients_path = raw_dir / "Inpatients_in_Health_Sectors_Hospitals__KSA.csv"
workforce_path = raw_dir / "PHC_Health_Workforce__MoH___Last_5_Years.csv"

inpatients_df = pd.read_csv(inpatients_path)
workforce_df = pd.read_csv(workforce_path)

print("Inpatients shape:", inpatients_df.shape)
print("Workforce shape:", workforce_df.shape)
```

## 3. Initial Data Inspection

Inspect dataset dimensions, columns, data types, and records.

```python
print("Inpatients columns:")
print(list(inpatients_df.columns))
display(inpatients_df)

print("Workforce columns:")
print(list(workforce_df.columns))
display(workforce_df)
```

```python
print("Inpatients information:")
inpatients_df.info()

print("\nWorkforce information:")
workforce_df.info()
```

## 4. Data Quality Checks

Check missing values, duplicate rows, and the structure of workforce categories.

```python
print("Missing values in Inpatients:")
print(inpatients_df.isna().sum())

print("\nMissing values in Workforce:")
print(workforce_df.isna().sum())

print("\nDuplicate rows in Inpatients:")
print(inpatients_df.duplicated().sum())

print("\nDuplicate rows in Workforce:")
print(workforce_df.duplicated().sum())
```

```python
print("Category frequency:")
print(workforce_df["Category"].value_counts(dropna=False))

print("\nCategory and nationality structure:")
display(workforce_df[["Category", "Nationality"]])
```

## 5. Audit Findings

- Two raw CSV datasets were successfully located and loaded.
- The inpatients dataset contains 5 rows and 6 columns.
- The workforce dataset contains 24 rows and 7 columns.
- No duplicate rows were detected in either dataset.
- No missing values were found in the inpatients dataset.
- The workforce `Category` column contains 16 missing values caused by the grouped table layout.
- All columns were loaded as strings and the year columns require numeric conversion.
- Some text labels in both datasets contain leading or trailing spaces.
- The inpatients dataset contains sector counts, a pre-calculated total, and a rate row with a different measurement unit.
- Both datasets contain year columns labeled from 2020G to 2024G.
- The datasets have different scopes: the inpatients data covers hospital sectors, while the workforce data covers MoH primary healthcare centers. Any direct ratio between them must be interpreted cautiously as a proxy indicator.
- The original raw files remain unchanged.
